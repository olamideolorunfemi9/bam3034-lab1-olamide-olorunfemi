## Part A BMI Analysis

The program reads each height and weight pair from `hw.txt`, converts cm to m and lb to kg, calculates BMI, then groups the results as Under Weight, Normal, Over Weight, or Obesity.

```python
"""BAM 3034 - Lab 1: Python Refresher
Student: Olamide Olorunfemi

Part A reads height/weight records from hw.txt, calculates BMI, then groups
people by BMI category. Part B demonstrates the required text-data functions.
"""

from pathlib import Path
import re


# ---------------------------------------------------------------------------
# Part A - BMI analysis from hw.txt
# ---------------------------------------------------------------------------
def load_height_weight_records(filename="hw.txt"):
    """Return a list of (height_cm, weight_lb) records from the input file."""
    values = [float(line.strip()) for line in Path(filename).read_text().splitlines()
              if line.strip()]
    if len(values) % 2 != 0:
        raise ValueError("hw.txt must have one height and one weight for each person.")
    return list(zip(values[0::2], values[1::2]))


def calculate_bmi(height_cm, weight_lb):
    """Calculate BMI after converting centimetres to metres and pounds to kg."""
    height_m = height_cm / 100
    weight_kg = weight_lb / 2.2
    return weight_kg / (height_m ** 2)


def bmi_category(bmi):
    """Return the BMI category required in the lab instructions."""
    if bmi < 18.5:
        return "Under Weight"
    if bmi < 25:
        return "Normal"
    if bmi < 30:
        return "Over Weight"
    return "Obesity"


def analyse_bmi_file(filename="hw.txt"):
    records = load_height_weight_records(filename)
    groups = {"Under Weight": [], "Normal": [], "Over Weight": [], "Obesity": []}

    for person_number, (height_cm, weight_lb) in enumerate(records, start=1):
        bmi = calculate_bmi(height_cm, weight_lb)
        groups[bmi_category(bmi)].append((person_number, bmi))

    for category, people in groups.items():
        print(f"{category}:")
        for person_number, bmi in people:
            print(f"Person- {person_number} : {bmi:.4f}")
        print(f"Group Size: {len(people)}")
        if people:
            group_average = sum(bmi for _, bmi in people) / len(people)
            print(f"Average BMI for this group: {group_average:.4f}")
        print()

    all_bmis = [bmi for people in groups.values() for _, bmi in people]
    print(f"The file has records for: {len(records)} persons")
    print(f"Average BMI for all these {len(records)} persons: {sum(all_bmis) / len(all_bmis):.4f}")
    return groups


```

```python
analyse_bmi_file("hw.txt")
```
